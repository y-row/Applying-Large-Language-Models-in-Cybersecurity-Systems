# **Part 12 — Practice: Checking Timeline**

In this section, you will practice a **simple timeline checking**:  from system log data.

Attack timeline reconstruction is a common task in **digital forensics and incident response**, where analysts examine logs to understand:

- When the attack started
- What actions the attacker performed
- Which systems were accessed
- The sequence of events during the attack



### Scenario

A security team detected suspicious activity on a server.  
You are given a **simplified log dataset** containing system events.

Your task is to:

1. Load the log data
2. Sort events by timestamp
3. Identify suspicious activities
4. Reconstruct a simple attack timeline


### Step 1 — Create a Sample Log Dataset

Run the following code to simulate system logs.

```python
import pandas as pd

log_data = {
    "timestamp": [
        "2025-03-10 09:00:12",
        "2025-03-10 09:05:33",
        "2025-03-10 09:10:02",
        "2025-03-10 09:12:45",
        "2025-03-10 09:20:11",
        "2025-03-10 09:35:40"
    ],
    "event": [
        "User login",
        "Failed login attempt",
        "Failed login attempt",
        "User login success",
        "File download",
        "Privilege escalation"
    ],
    "user": [
        "admin",
        "unknown",
        "unknown",
        "admin",
        "admin",
        "admin"
    ]
}

logs = pd.DataFrame(log_data)
logs
```

Step 2 — Convert Timestamp Format

Convert the timestamp column into a proper datetime format.

```python
logs["timestamp"] = pd.to_datetime(logs["timestamp"])
logs
```

Step 3 — Sort Events by Time

Reconstruct the timeline by sorting the events.

```python
timeline = logs.sort_values(by="timestamp")
timeline
```

Step 4 — Checking the failed login attempts, privilege escalation and unusual downloads

Suspicious activities might include......

```python
ourfocus = logs[logs["event"].str.contains("Failed|Privilege", case=False)]
ourfocus
```

Step 5 — Visualize the Attack Timeline

Display the ordered timeline.

```python
for index, row in timeline.iterrows():
    print(row["timestamp"], "-", row["event"], "-", row["user"])
```

# Local Service Installation Threat Hunting

This notebook demonstrates how threat hunting can be performed to detect suspicious local service installation activity on Windows systems. The hunt is based on the Threat Hunter Playbook methodology and focuses on identifying adversaries creating new Windows services to execute malicious code or maintain persistence within an environment.

## Reference

- Threat Hunter Playbook – Local Service Installation  
  https://threathunterplaybook.com/hunts/windows/190813-LocalServiceInstallation/notebook.html

# Threat Hunting Hypothesis

Adversaries may create new Windows services to execute malicious binaries, commands, or scripts on compromised endpoints. Attackers often abuse the Windows Service Control Manager (SCM) to establish persistence, escalate privileges, or execute payloads during lateral movement operations.

This technique is commonly associated with:
- Persistence
- Privilege Escalation
- Lateral Movement

under the MITRE ATT&CK framework.

---

# Dataset Overview

The dataset used in this notebook contains Windows event logs generated from simulated adversary activity involving remote service creation using PsExec and DCERPC methods.

The dataset includes telemetry from:
- Windows Security Logs
- Sysmon Logs
- Service installation events
- Authentication events

Download & Extract Dataset

```python
import requests
from zipfile import ZipFile
from io import BytesIO

# Dataset URL
url = "https://raw.githubusercontent.com/OTRF/Security-Datasets/master/datasets/atomic/windows/lateral_movement/host/empire_psexec_dcerpc_tcp_svcctl.zip"

# Download ZIP dataset
zipFileRequest = requests.get(url)

# Extract ZIP file
zipFile = ZipFile(BytesIO(zipFileRequest.content))

# Extract JSON dataset
datasetJSONPath = zipFile.extract(zipFile.namelist()[0])

print("Dataset extracted successfully:")
print(datasetJSONPath)
```

Read JSON Dataset

```python
import pandas as pd
from pandas.io import json

# Read JSON logs
df = json.read_json(
    path_or_buf=datasetJSONPath,
    lines=True
)

# Display dataset shape
print(df.shape)

# Preview first rows
df.head()
```

Explore Available Channels

```python
df['Channel'].value_counts()
```

Explore Event IDs

```python
df['EventID'].value_counts().head(20)
```

# Analytic I — Local Service Installation

Look for new services being created in the environment.

```python
analytic_1 = (
    df[
        [
            '@timestamp',
            'Hostname',
            'SubjectUserName',
            'ServiceName',
            'ServiceType',
            'ServiceStartType',
            'ServiceAccount'
        ]
    ]
    [
        (df['Channel'].str.lower() == 'security')
        & (df['EventID'] == 4697)
    ]
)

analytic_1.head(10)
```

This analytic identifies Windows Security Event ID 4697, which indicates a new service installation.

# Analytic II — Remote Service Creation

Detect remote service creation associated with network logon sessions.

```python
# Service installation events
serviceInstallDf = (
    df[
        [
            '@timestamp',
            'Hostname',
            'SubjectUserName',
            'SubjectLogonId',
            'ServiceName',
            'ServiceType'
        ]
    ]
    [
        (df['Channel'].str.lower() == 'security')
        & (df['EventID'] == 4697)
    ]
)

# Network logon events
networkLogonDf = (
    df[
        [
            '@timestamp',
            'Hostname',
            'TargetUserName',
            'TargetLogonId',
            'IpAddress',
            'LogonType'
        ]
    ]
    [
        (df['Channel'].str.lower() == 'security')
        & (df['EventID'] == 4624)
        & (df['LogonType'] == 3)
        & (~df['TargetUserName'].str.endswith('$', na=False))
    ]
)

# Correlate service creation with network logon
remote_service_creation = pd.merge(
    serviceInstallDf,
    networkLogonDf,
    left_on='SubjectLogonId',
    right_on='TargetLogonId',
    how='inner'
)

remote_service_creation.head(10)
```

Export Results to CSV

```python
remote_service_creation.to_csv(
    "remote_service_creation_results.csv",
    index=False
)

print("Results exported successfully.")
```
